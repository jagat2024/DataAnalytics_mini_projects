** Distribution plot is use to visualize the distribution of quantitative data **

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sls
```

```python
df=sls.load_dataset("tips")
df
```

```python
df["size"].unique()
```

```python
sls.displot(df["total_bill"])
```

```python
sls.histplot(df["total_bill"],kde=True)
```

```python
plt.subplot(1,2,1)
sls.histplot(df["total_bill"],kde=True)
plt.subplot(1,2,2)
sls.histplot(df["tip"],kde=True)
```

```python
sls.jointplot(x="total_bill",y="tip",data=df,kind="kde")
```

```python
sls.jointplot(x="total_bill",y="tip",data=df,kind="hist")
```

```python
sls.jointplot(x="total_bill",y="tip",data=df,kind="hex")
```

```python
sls.jointplot(x="total_bill",y="tip",data=df,kind="reg")
```

```python
sls.pairplot(df,hue="smoker",palette="rainbow")
```

```python
sls.rugplot(df)
```
