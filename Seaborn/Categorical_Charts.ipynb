# Categorial Dataplots

#Now lets discuss using seaborn to plot categirical data.There are few main plot types for this:
* boxplot
* Violinplot
* Stripplot
* Swarmplot
* Barplot
* Countplot


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=sns.load_dataset("tips")
df
```

```python
k=sns.countplot(x=df["sex"])
plt.show()
```

```python
k=sns.countplot(x=df["sex"],hue=df['smoker'])
plt.show()
```

```python
sns.barplot(x=df["total_bill"],y=df["tip"])
plt.show()
```

```python
sns.barplot(x=df["sex"],y=df["tip"],estimator=np.sum)
plt.show()
```

```python
sns.barplot(x=df["sex"],y=df["total_bill"],estimator=np.sum)
plt.show()
```

```python
sns.boxplot(x=df["tip"],y=df["day"],palette="rainbow")
plt.show()
```

# Boxplot and violin plot
Boxplot and violin plot used to show the distribution of the categorical data.A box plot shows the distribution of quantitative data in a way that facilates the comaprision between variables or across levels of a categorical variable. The box shows the quartile of the dataset while the whisker extend to show the rest of the distribution ,except the points that are determined to be "outliers"using a method that is a function of the inter quartile range 


```python
sns.violinplot(y=df["total_bill"],x=df["day"],palette="rainbow")
plt.show()
```

# strip plot
A Strip Plot is a data visualization technique used to display the distribution of individual data points across one or more categorical variables. It represents each observation as a dot along a numerical axis, helping identify patterns, spread, and outliers.

In Seaborn, it is created using the sns.stripplot() function.
* Parameter	Datatype	    Description
* data	    DataFrame	    Dataset containing the data
* x	        Categorical	    Category such as day, gender, department
* y	        Numerical	    Numeric values such as tip, salary, marks
* hue	        Categorical	    Additional grouping variable represented by colors
* jitter	    Boolean	        Controls the spreading of overlapping points
* orient	    String	        Defines plot orientation ("x" or "y" in applicable contexts)

```python
sns.stripplot(x="total_bill",y="day",data=df,hue="sex",jitter=True)
plt.show()
```
