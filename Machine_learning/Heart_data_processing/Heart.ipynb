```python
df['Cholesterol'].value_counts()#we can see that about 172 peope has 0 cholestrol which is practically not possible so we need to clean the data by not taking the vaalues in calculating the mean ans replacing the 0 with the mean value
```

```python
ch_mean=df.loc[df['Cholesterol']!=0,'Cholesterol'].mean()
print(ch_mean)
df["Cholesterol"]=df["Cholesterol"].replace(0,ch_mean)
df["Cholesterol"]=df["Cholesterol"].round(2)
df.head()
```

```python
Resting_bp_mean=df.loc[df["RestingBP"]!=0,"RestingBP"].mean()
print(Resting_bp_mean)

df["RestingBP"]=df["RestingBP"].replace(0,Resting_bp_mean)
df["RestingBP"]=df["RestingBP"].round(2)
df.head()
```

```python
def plotting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(df[var],kde=True)
plotting('Age',1)
plotting('RestingBP',2)
plotting('Cholesterol',3)
plotting('MaxHR',4)
#after making all the chnages the graphs are now looking as normal standardidation 
```

# Data Preprocessing


```python
df_encode=pd.get_dummies(df,drop_first=True)
df_encode.astype(int)
```

```python
from sklearn.preprocessing import StandardScaler
numeric_columns=['Age','RestingBP','Cholesterol','MaxHR','Oldpeak']
scaler=StandardScaler()
df_encode[numeric_columns]=scaler.fit_transform(df_encode[numeric_columns])
```

# Feature Engineering


```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 30, 40, 50, 60, 70, 120],
    labels=[
        "Under30", "30s", "40s",
        "50s", "60s", "70plus"
    ],
    right=False
)
df.head()
```
