```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
df=sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
#as we can see in deck there is a high number of null valiuue so we will drop that column
df.drop(["deck","embark_town","alive","class","parch","who","adult_male"],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df["age"].fillna(df["age"].mean(),inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df.dropna(subset=["embarked"],inplace=True)#subset dropna use kiya hai kyu ki wo un specific values ko remove kartihai matlab un rows ko jo kli NaN values rakhi hai

df.info()
```

```python
df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df["sex"]=le.fit_transform(df["sex"])
df["embarked"]=le.fit_transform(df["embarked"])
```

```python
df.head()
df.astype(int)
```

```python
x=df.drop("survived",axis=1)
y=df["survived"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train,y_train);
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
#in loigistic regression we use confusion matrix just like in linear regression we use adjusted R2 and R2
```

```python
from sklearn.metrics import confusion_matrix;
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
sns.heatmap(cm, annot=True, fmt="d",cmap="Blues")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
from sklearn.metrics import accuracy_score,f1_score,precision_score,classification_report # classification report kisi bhi classification algorithm keliye use kar sakte hain
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
f1_score(y_pred,y_test)
```

```python
precision_score(y_pred,y_test)
```

```python
from sklearn.metrics import log_loss

y_prob = model.predict_proba(X_test)

print("Log Loss:", log_loss(y_test, y_prob))
```

```python
#Feature scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler();
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.fit_transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=5)

```

```python
knn_model.fit(X_train_scaled,y_train)
```

```python
y_knn_pred=knn_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import classification_report
accuracy_score(y_test,y_knn_pred)          #accuracy of knn model is less than logistic regression
```

```python
confusion_matrix(y_test,y_knn_pred)
```

```python
print(classification_report(y_test,y_knn_pred))
```
