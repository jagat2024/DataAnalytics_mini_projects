```python
import numpy as np

# 1. Create a 1D array from a Python list
arr = np.array([1, 2, 3, 4, 5])
print("Original Array:", arr)

# 2. Perform a fast element-wise operation (multiply every item by 2)
doubled_arr = arr * 2
print("Doubled Array: ", doubled_arr)

# 3. Calculate the sum of all elements
total_sum = np.sum(arr)
print("Sum of elements:", total_sum)
```
