# two sample t-test

```python
import numpy as np
from scipy import stats
```

```python
group_A=[85,88,90,92,87,85,89,91,86,88]
group_B=[82,84,80,83,81,79,78,85,84,83]
t_stats,p_val=stats.ttest_ind(group_A,group_B,equal_var=False)
alpha=0.05
if p_val<alpha:
    print("reject")
else:
    print("accept")
```

```python
import seaborn as sns
import pandas as pd
from scipy.stats import f_oneway
df=sns.load_dataset("titanic")
df.isna(0)
```

```python
df["pclass"].unique()
```

```python
class1 = df.loc[(df["pclass"] == 1) & df["age"].notna(), "age"]

class2 = df.loc[(df["pclass"] == 2) & df["age"].notna(), "age"]

class3 = df.loc[(df["pclass"] == 3) & df["age"].notna(), "age"]
f_stats,p_val=f_oneway(class1,class2,class3)

print(p_val)
```

```python
alpha=0.05
if(p_val<alpha):
    print("reject the null hypothesis")
else:
    print("accept the null hypothesis")
```
