# z-test using  p-test


```python
import numpy as np
from scipy.stats import norm
```

```python
sample=[172,174,168,169,171,173,175,170,169,172]
population_mean=170
population_standard=3
sample_mean=np.mean(sample)
n=len(sample)
z_score=(sample_mean - population_mean)/(population_standard/np.sqrt(n))
p_value=2*(1-norm.cdf(abs(z_score)))
print(p_value)
```

```python
alpha=0.05
if(p_value<alpha):
    print("reject the null hypothesis")
else:
    print("accept the null hypothesis")
    
```

# T-test by P-test

```python
import numpy as np 
from scipy import stats
sample=[172,174,168,169,171,173,175,170,169,172]
sample_std=np.std(sample,ddof=1)
sample_mean=np.mean(sample)
n=len(sample)
population_mean=170
t_stat=(sample_mean-population_mean)/(sample_std/np.sqrt(n))
print(t_stat)
```

```python
p_value=2*(1-stats.t.cdf(abs(t_stat),df=n-1))
alpha=0.05
if(p_value<alpha):
    print("reject")
else:
    print("accept")
```
