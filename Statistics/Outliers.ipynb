# outliers summary and there use case

```python
import numpy as np 
arr=np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

```python
q1=np.percentile(arr,25)
print(q1)
```

```python
q3=np.percentile(arr,75)
print(q3)
```

```python
IQR=q3-q1
print(IQR)
```

```python
uf=q3+1.5*(IQR)
lf=q3-1.5*(IQR)
```

```python
print(uf,lf)
```

```python
l=[]
for i in arr:
    if i>=lf and i<=uf :
        l.append(i)
l
arr2=np.array(l)
```
